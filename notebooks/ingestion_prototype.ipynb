## 3. Cross-sample consistency check

Before building the ingestion pipeline, we need to understand how stable the data schema is across **different states** and **time periods**. DATASUS has been collecting data since the mid-90s, and the SIM system has gone through multiple versions — so we should expect potential differences in:
- Available columns (fields added or removed over time)
- Data types (same field stored differently across years)
- Data completeness (older data may have more gaps)

We'll fetch samples spanning **1996–2022** from different regions to stress-test this.

```python
# Define samples: different regions and time periods
samples = {
    "SP_2022": {"states": ["SP"], "years": [2022]},  # Southeast, recent (already loaded)
    "AM_2022": {"states": ["AM"], "years": [2022]},  # North region, recent
    "BA_2022": {"states": ["BA"], "years": [2022]},  # Northeast, recent
    "SP_2010": {"states": ["SP"], "years": [2010]},  # Southeast, mid-period
    "MG_2000": {"states": ["MG"], "years": [2000]},  # Southeast, early 2000s
    "SP_1996": {"states": ["SP"], "years": [1996]},  # Southeast, oldest available CID10
}

dataframes = {"SP_2022": df}  # reuse the one we already have

for name, params in samples.items():
    if name == "SP_2022":
        continue
    print(f"Downloading {name}...")
    raw_sample = download(states=params["states"], years=params["years"], groups=["CID10"])
    dataframes[name] = raw_sample.to_dataframe()
    print(f"  {name}: {dataframes[name].shape}")

print("\nAll samples loaded.")
```

### 3.2 Column comparison

Using SP_2022 as reference, check which columns are missing or extra in each sample. This tells us what our pipeline needs to handle — e.g., should we enforce a fixed schema and fill missing columns with nulls?

```python
# Check if all samples have the same columns
reference_cols = set(dataframes["SP_2022"].columns)

for name, d in dataframes.items():
    current_cols = set(d.columns)
    missing = reference_cols - current_cols
    extra = current_cols - reference_cols
    
    status = "OK" if not missing and not extra else "MISMATCH"
    print(f"{name}: {d.shape[1]} cols — {status}")
    if missing:
        print(f"  Missing vs SP_2022: {missing}")
    if extra:
        print(f"  Extra vs SP_2022:   {extra}")
```

### 3.3 Dtype consistency

Even when columns exist in all samples, they may be stored with different types (e.g., `Int64` vs `string`). This matters for BigQuery schema definition and for avoiding silent type coercion bugs in the pipeline.

```python
# Compare dtype consistency per column across samples
dtype_comparison = pd.DataFrame({
    name: d.dtypes for name, d in dataframes.items()
})
dtype_comparison["all_same"] = dtype_comparison.apply(lambda row: row.nunique() == 1, axis=1)

# Show only columns where dtypes differ across samples
mismatches = dtype_comparison[~dtype_comparison["all_same"]]
if mismatches.empty:
    print("All columns have consistent dtypes across all samples.")
else:
    print(f"Dtype mismatches found in {len(mismatches)} columns:")
    print(mismatches.drop(columns="all_same"))
```

### 3.4 Blank/null rates for key fields

Not all "non-null" values are meaningful — DATASUS often uses empty strings instead of proper nulls. Here we check the **effective fill rate** of the fields we care about most for the pipeline. High blank rates in older data may affect how far back we can reliably analyze trends.

```python
# Compare empty/blank rates across samples for key fields
key_fields = ["CAUSABAS", "SEXO", "IDADE", "RACACOR", "CODMUNRES", "DTOBITO", "DTNASC", "LOCOCOR", "ESC2010","ESC"]

for field in key_fields:
    print(f"\n--- {field} ---")
    for name, d in dataframes.items():
        if field not in d.columns:
            print(f"  {name}: COLUMN MISSING")
            continue
        total = len(d)
        blank = (d[field].astype(str).str.strip() == "").sum()
        null = d[field].isnull().sum()
        unique = d[field].nunique()
        print(f"  {name}: {total:>8} rows | {blank:>6} blank ({blank/total*100:.1f}%) | {null:>5} null | {unique:>5} unique")
```

### 3.5 R99 quality check — ill-defined causes with filled certificate lines

`R99` ("Other ill-defined and unspecified causes of mortality") is a catch-all code that signals the underlying cause couldn't be determined. However, if the certificate lines (LINHAA–LINHAD, LINHAII) still contain ICD-10 codes, it may indicate a coding/selection issue rather than truly unknown cause. A high rate of this pattern — especially in older data — is a data quality signal worth monitoring in the pipeline.

```python
# Check across all samples: CAUSABAS=R99 (ill-defined) but LINHAs have data
linhas = ["LINHAA", "LINHAB", "LINHAC", "LINHAD", "LINHAII"]

for name, d in dataframes.items():
    causabas_r99 = d["CAUSABAS"].astype(str).str.strip() == "R99"
    any_linha_filled = d[linhas].astype(str).apply(lambda col: col.str.strip() != "").any(axis=1)
    
    total_r99 = causabas_r99.sum()
    r99_with_linhas = (causabas_r99 & any_linha_filled).sum()
    pct = (r99_with_linhas / total_r99 * 100) if total_r99 > 0 else 0
    
    print(f"{name}: {total_r99:>6} R99 total | {r99_with_linhas:>6} with LINHA data ({pct:.1f}%)")
```

### 4.1 Empty string vs null analysis

DATASUS encodes missing data as empty strings rather than SQL-style nulls. We count **effective blanks** (empty + whitespace-only) vs true Python/numpy nulls for every column in every sample, then visualize the blank-rate matrix as a heatmap.

```python
# Compute effective blank % (empty + whitespace-only + null) per column per sample
# Use the union of all columns across samples
all_cols = sorted(set().union(*(d.columns for d in dataframes.values())))

blank_pct = pd.DataFrame(index=all_cols, columns=list(dataframes.keys()), dtype=float)

for name, d in dataframes.items():
    total = len(d)
    for col in all_cols:
        if col not in d.columns:
            blank_pct.loc[col, name] = np.nan  # column doesn't exist in this sample
        else:
            is_null = d[col].isnull()
            is_blank = d[col].astype(str).str.strip() == ""
            effective_blank = (is_null | is_blank).sum()
            blank_pct.loc[col, name] = effective_blank / total * 100

# Show summary: how many columns are >50% blank per sample
print("Columns with >50% effective blanks per sample:")
print((blank_pct > 50).sum())
print(f"\nTotal columns across all samples: {len(all_cols)}")
```

```python
# Heatmap: blank % by column x sample
# Filter to columns that have at least some blanks in at least one sample (skip always-0 cols)
has_blanks = blank_pct.loc[(blank_pct > 0).any(axis=1)].copy()

fig, ax = plt.subplots(figsize=(10, max(8, len(has_blanks) * 0.25)))

cmap = matplotlib.colormaps["YlOrRd"].copy()
cmap.set_bad(color="lightgray")  # NaN = column missing → gray

im = ax.imshow(has_blanks.values.astype(float), aspect="auto", cmap=cmap, vmin=0, vmax=100)

ax.set_xticks(range(len(has_blanks.columns)))
ax.set_xticklabels(has_blanks.columns, rotation=45, ha="right")
ax.set_yticks(range(len(has_blanks.index)))
ax.set_yticklabels(has_blanks.index, fontsize=7)
ax.set_xlabel("Sample")
ax.set_ylabel("Column")
ax.set_title("Effective Blank % by Column × Sample\n(gray = column missing in sample)")

plt.colorbar(im, ax=ax, label="Blank %", shrink=0.6)
plt.tight_layout()
plt.show()
```

### 4.2 Key field validity checks

For our core pipeline fields, check for invalid/unexpected values based on domain rules:

| Field | Valid values |
|-------|------------|
| `SEXO` | 0 (ignored), 1 (male), 2 (female) |
| `RACACOR` | 1–5 (branca, preta, amarela, parda, indígena) |
| `LOCOCOR` | 1–6, 9 (hospital, other health, home, street, other, unknown) |
| `IDADE` | First digit 0–5 encodes age unit (minutes, hours, days, months, years, 100+years) |
| `DTOBITO` | 8-digit date string DDMMYYYY |
| `CAUSABAS` | ICD-10 format: letter + 2-3 digits (e.g. I219, R99) |

```python
# Validity check functions for each core field
def check_sexo(series):
    """SEXO should be in {0, 1, 2} or blank."""
    vals = series.astype(str).str.strip()
    filled = vals[vals != ""]
    return (~filled.isin(["0", "1", "2"])).sum(), len(filled)

def check_racacor(series):
    """RACACOR should be in {1, 2, 3, 4, 5} or blank."""
    vals = series.astype(str).str.strip()
    filled = vals[vals != ""]
    return (~filled.isin(["1", "2", "3", "4", "5"])).sum(), len(filled)

def check_lococor(series):
    """LOCOCOR should be in {1, 2, 3, 4, 5, 6, 9} or blank."""
    vals = series.astype(str).str.strip()
    filled = vals[vals != ""]
    return (~filled.isin(["1", "2", "3", "4", "5", "6", "9"])).sum(), len(filled)

def check_idade(series):
    """IDADE first digit should be 0-5 (age encoding unit)."""
    vals = series.astype(str).str.strip()
    filled = vals[vals != ""]
    first_digit = filled.str[0]
    return (~first_digit.isin(["0", "1", "2", "3", "4", "5"])).sum(), len(filled)

def check_dtobito(series):
    """DTOBITO should be 8-digit date string (DDMMYYYY)."""
    vals = series.astype(str).str.strip()
    filled = vals[vals != ""]
    valid_pattern = filled.str.match(r"^\d{8}$")
    return (~valid_pattern).sum(), len(filled)

def check_causabas(series):
    """CAUSABAS should match ICD-10 pattern: letter + 2-4 chars."""
    vals = series.astype(str).str.strip()
    filled = vals[vals != ""]
    valid_pattern = filled.str.match(r"^[A-Z]\d{2,4}$")
    return (~valid_pattern).sum(), len(filled)

# Run validity checks across all samples
validity_checks = {
    "SEXO": check_sexo,
    "RACACOR": check_racacor,
    "LOCOCOR": check_lococor,
    "IDADE": check_idade,
    "DTOBITO": check_dtobito,
    "CAUSABAS": check_causabas,
}

results = []
for name, d in dataframes.items():
    for field, check_fn in validity_checks.items():
        if field not in d.columns:
            results.append({"sample": name, "field": field, "invalid": None, "filled": None, "invalid_pct": None})
        else:
            inv, filled = check_fn(d[field])
            results.append({
                "sample": name, "field": field,
                "invalid": inv, "filled": filled,
                "invalid_pct": round(inv / filled * 100, 2) if filled > 0 else 0
            })

validity_df = pd.DataFrame(results)
# Pivot for readability: field x sample
pivot = validity_df.pivot(index="field", columns="sample", values="invalid_pct")
print("Invalid value % by field × sample:")
print(pivot.to_string())
print("\n--- Absolute invalid counts ---")
pivot_abs = validity_df.pivot(index="field", columns="sample", values="invalid")
print(pivot_abs.to_string())
```

```python
# Show examples of invalid values for any field with >0% invalid
print("Examples of invalid values found:")
for name, d in dataframes.items():
    for field, check_fn in validity_checks.items():
        if field not in d.columns:
            continue
        inv, filled = check_fn(d[field])
        if inv > 0:
            vals = d[field].astype(str).str.strip()
            filled_vals = vals[vals != ""]
            # Get the invalid ones
            if field == "SEXO":
                invalid_mask = ~filled_vals.isin(["0", "1", "2"])
            elif field == "RACACOR":
                invalid_mask = ~filled_vals.isin(["1", "2", "3", "4", "5"])
            elif field == "LOCOCOR":
                invalid_mask = ~filled_vals.isin(["1", "2", "3", "4", "5", "6", "9"])
            elif field == "IDADE":
                invalid_mask = ~filled_vals.str[0].isin(["0", "1", "2", "3", "4", "5"])
            elif field == "DTOBITO":
                invalid_mask = ~filled_vals.str.match(r"^\d{8}$")
            elif field == "CAUSABAS":
                invalid_mask = ~filled_vals.str.match(r"^[A-Z]\d{2,4}$")
            else:
                continue
            invalid_examples = filled_vals[invalid_mask].value_counts().head(5)
            if len(invalid_examples) > 0:
                print(f"\n  {name} / {field} ({inv} invalid): {dict(invalid_examples)}")
```

### 4.3 Completeness score per sample

For each row, compute the percentage of **core fields** (`DTOBITO`, `CAUSABAS`, `SEXO`, `IDADE`, `RACACOR`, `CODMUNRES`, `LOCOCOR`) that are both present and valid. Then compare the distribution across samples to see if older years are significantly worse.

```python
core_fields = ["DTOBITO", "CAUSABAS", "SEXO", "IDADE", "RACACOR", "CODMUNRES", "LOCOCOR"]

completeness_stats = {}

for name, d in dataframes.items():
    available_core = [f for f in core_fields if f in d.columns]
    n_core = len(core_fields)

    # For each row, count how many core fields are non-blank
    filled_counts = pd.Series(0, index=d.index)
    for f in available_core:
        is_filled = (d[f].astype(str).str.strip() != "") & d[f].notna()
        filled_counts += is_filled.astype(int)

    completeness_pct = filled_counts / n_core * 100
    completeness_stats[name] = {
        "mean": completeness_pct.mean(),
        "median": completeness_pct.median(),
        "min": completeness_pct.min(),
        "p10": completeness_pct.quantile(0.10),
        "p25": completeness_pct.quantile(0.25),
        "100%_complete": (completeness_pct == 100).mean() * 100,
        "available_core_fields": len(available_core),
    }

completeness_summary = pd.DataFrame(completeness_stats).T.round(2)
print("Core field completeness per sample (% of 7 core fields filled per row):")
print(completeness_summary.to_string())
```

```python
# Completeness distribution per sample — box plot
fig, ax = plt.subplots(figsize=(10, 5))

completeness_data = []
labels = []
for name, d in dataframes.items():
    available_core = [f for f in core_fields if f in d.columns]
    filled_counts = pd.Series(0, index=d.index)
    for f in available_core:
        is_filled = (d[f].astype(str).str.strip() != "") & d[f].notna()
        filled_counts += is_filled.astype(int)
    completeness_data.append((filled_counts / len(core_fields) * 100).values)
    labels.append(name)

ax.boxplot(completeness_data, labels=labels, vert=True)
ax.set_ylabel("Completeness %")
ax.set_title("Per-row completeness of core fields by sample")
ax.set_xticklabels(labels, rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### 4.5 Observations & recommendations

**Key findings from the data quality deep-dive:**

1. **Empty strings vs nulls**: DATASUS consistently uses empty strings for missing data, not nulls. The ingestion layer must treat `""` and whitespace-only strings as null equivalents. A `COALESCE(NULLIF(TRIM(field), ''), NULL)` pattern in SQL or equivalent pandas logic is required.

2. **Field validity**:
   - `SEXO`, `LOCOCOR`, `RACACOR`: Generally clean with valid domain values, but blank rates vary significantly by year and region (especially `RACACOR` in older years).
   - `IDADE`: Encoding is consistent (first digit = unit), but older data may have edge cases. The pipeline should validate the first digit and flag unexpected values.
   - `DTOBITO`: Format is stable (8-digit DDMMYYYY) across all years. Safe to parse directly.
   - `CAUSABAS`: ICD-10 format is consistent. Any non-matching values are rare and can be flagged as quality issues.

3. **Completeness**:
   - Recent years (2022) have significantly higher completeness than older years (1996, 2000).
   - `RACACOR` is the most problematic core field — often blank in older data (race/ethnicity collection improved over time).
   - All samples have the critical fields `DTOBITO`, `CAUSABAS`, `SEXO`, and `CODMUNRES` well-populated.

4. **Column availability**:
   - Several columns were added in later SIM versions and are absent in older files (e.g., `ESC2010`, `ESTCIV`, newer audit fields).
   - All 7 core pipeline fields are available across all sampled years (1996–2022) — **the pipeline can safely target all years from 1996 onward** for the core schema.

**Recommendations for the ingestion layer:**
- **Blanks → NULL**: Apply blank-to-null normalization on all string columns at ingestion time.
- **RACACOR fallback**: For years before ~2005, expect high blank rates for `RACACOR`. Use a default/unknown category rather than dropping rows.
- **Schema union**: Use the union of all columns across years, with missing columns filled as NULL. The dbt layer should handle this with `SAFE_CAST` and defaults.
- **Validation gate**: Implement quality checks as a dbt test or post-ingestion step — flag rows where core fields fail validity rules rather than dropping them.
- **Year range**: All years from 1996 onward are viable for ingestion. Pre-1996 data uses CID-9 (ICD-9) and would need a separate mapping layer.
