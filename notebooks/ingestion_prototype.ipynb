## Data Dictionary — SIM-DO (Declaração de Óbito)

Source: [DATASUS/SVS](https://svs.aids.gov.br/download/Dicionario_de_Dados_SIM_tabela_DO.pdf) | [PCDaS/Fiocruz](https://pcdas.icict.fiocruz.br/conjunto-de-dados/sistema-de-informacoes-de-mortalidade-sim/dicionario-de-variaveis/)

### Identificação e Dados Administrativos

| Campo | Descrição |
|-------|-----------|
| NUMERODO | Número da Declaração de Óbito |
| ORIGEM | Origem do registro |
| CONTADOR | Identificador do registro |
| NUMEROLOTE | Número do lote de processamento |
| VERSAOSIST | Versão do sistema SIM |
| VERSAOSCB | Versão do Seletor de Causa Básica (SCB) |
| CODESTAB | Código CNES do estabelecimento de saúde |
| ESTABDESCR | Descrição do estabelecimento |

### Tipo e Data do Óbito

| Campo | Descrição |
|-------|-----------|
| TIPOBITO | Tipo do óbito: 1=Fetal; 2=Não fetal |
| DTOBITO | Data do óbito (ddmmaaaa) |
| HORAOBITO | Hora do óbito |

### Dados do Falecido

| Campo | Descrição |
|-------|-----------|
| NATURAL | Naturalidade (1º dígito=8 + código UF para brasileiros) |
| CODMUNNATU | Código IBGE do município de naturalidade |
| DTNASC | Data de nascimento (ddmmaaaa) |
| IDADE | Idade codificada — 1º dígito: 0=min, 1=horas, 2=dias, 3=meses, 4=anos, 5=100+ anos |
| SEXO | Sexo: 0=Ignorado; 1=Masculino; 2=Feminino |
| RACACOR | Raça/Cor: 1=Branca; 2=Preta; 3=Amarela; 4=Parda; 5=Indígena |
| ESTCIV | Estado civil: 1=Solteiro; 2=Casado; 3=Viúvo; 4=Separado; 5=União consensual; 9=Ignorado |
| ESC | Escolaridade: 1=Nenhuma; 2=1-3 anos; 3=4-7; 4=8-11; 5=12+; 9=Ignorado |
| ESC2010 | Escolaridade (2010): 0=Sem; 1=Fund I; 2=Fund II; 3=Médio; 4=Sup incompleto; 5=Sup completo; 9=Ignorado |
| OCUP | Ocupação (CBO-2002) |

### Residência e Local de Ocorrência

| Campo | Descrição |
|-------|-----------|
| CODMUNRES | Código IBGE do município de residência (6 dígitos) |
| LOCOCOR | Local de ocorrência: 1=Hospital; 2=Outro estab. saúde; 3=Domicílio; 4=Via pública; 5=Outros; 6=Aldeia indígena; 9=Ignorado |
| CODMUNOCOR | Código IBGE do município de ocorrência |

### Dados Maternos / Fetais / Perinatais

| Campo | Descrição |
|-------|-----------|
| IDADEMAE | Idade da mãe (anos) |
| ESCMAE / ESCMAE2010 | Escolaridade da mãe |
| OCUPMAE | Ocupação da mãe (CBO-2002) |
| QTDFILVIVO | Nº filhos vivos |
| QTDFILMORT | Nº filhos mortos |
| GRAVIDEZ | Tipo: 1=Única; 2=Dupla; 3=Tripla+; 9=Ignorado |
| GESTACAO / SEMAGESTAC | Semanas de gestação (faixas / valor numérico) |
| PARTO | Tipo: 1=Vaginal; 2=Cesáreo; 9=Ignorado |
| OBITOPARTO | Óbito em relação ao parto: 1=Antes; 2=Durante; 3=Depois; 9=Ignorado |
| PESO | Peso ao nascer (gramas) |
| OBITOGRAV | Óbito na gravidez: 1=Sim; 2=Não; 9=Ignorado |
| OBITOPUERP | Óbito no puerpério: 1=0-42 dias; 2=43 dias-1 ano; 3=Não; 9=Ignorado |
| TPMORTEOCO | Momento da morte: 1=Gravidez; 2=Parto; 3=Aborto; 4=Até 42d pós-parto; 5=43d-1 ano; 8=Fora destes; 9=Ignorado |

### Assistência Médica e Confirmação

| Campo | Descrição |
|-------|-----------|
| ASSISTMED | Assistência médica: 1=Sim; 2=Não; 9=Ignorado |
| EXAME | Exame complementar: 1=Sim; 2=Não; 9=Ignorado |
| CIRURGIA | Houve cirurgia: 1=Sim; 2=Não; 9=Ignorado |
| NECROPSIA | Houve necropsia: 1=Sim; 2=Não; 9=Ignorado |
| ATESTANTE | Tipo: 1=Médico assistente; 2=Substituto; 3=IML; 4=SVO; 5=Outro |

### Causa do Óbito

| Campo | Descrição |
|-------|-----------|
| LINHAA / LINHAB / LINHAC / LINHAD | Linhas A–D do atestado (códigos CID-10) |
| LINHAII | Parte II — outras condições significativas |
| CAUSABAS | **Causa básica do óbito (CID-10)** — campo principal |
| CAUSABAS_O | Causa básica original (antes de re-seleção) |
| CB_PRE | Causa selecionada sem re-seleção |
| CAUSAMAT | Causa externa associada a morte materna |

### Causas Externas

| Campo | Descrição |
|-------|-----------|
| CIRCOBITO | Circunstância: 1=Acidente; 2=Suicídio; 3=Homicídio; 4=Outros; 9=Ignorado |
| ACIDTRAB | Acidente de trabalho: 1=Sim; 2=Não; 9=Ignorado |
| FONTE | Fonte: 1=BO; 2=Hospital; 3=Família; 4=Outra; 9=Ignorado |

### Investigação

| Campo | Descrição |
|-------|-----------|
| TPPOS | Investigação: 1=Investigado; 2=Não investigado |
| DTINVESTIG | Data da investigação |
| FONTEINV | Fonte da investigação (1-9) |

### Processamento

| Campo | Descrição |
|-------|-----------|
| STCODIFICA | Codificado em estação: S/N |
| CODIFICADO | Registro codificado: S/N |
| DTATESTADO / DTCADASTRO / DTRECEBIM / DTRECORIGA | Datas de processamento |
| DIFDATA | Diferença (dias) entre óbito e recebimento |
| STDOEPIDEM | DO epidemiológica: 1=Sim; 0=Não |
| STDONOVA | DO nova (substitutiva): 1=Sim; 0=Não |

## Data Dictionary — SIM-DO (Death Certificate) — English Version

Source: [DATASUS/SVS](https://svs.aids.gov.br/download/Dicionario_de_Dados_SIM_tabela_DO.pdf) | [PCDaS/Fiocruz](https://pcdas.icict.fiocruz.br/conjunto-de-dados/sistema-de-informacoes-de-mortalidade-sim/dicionario-de-variaveis/)

### Identification & Administrative Fields

| Field | Description |
|-------|-------------|
| NUMERODO | Death Certificate number |
| ORIGEM | Record origin |
| CONTADOR | Record identifier |
| NUMEROLOTE | Processing batch number |
| VERSAOSIST | SIM system version |
| VERSAOSCB | Underlying Cause Selector (SCB) version |
| CODESTAB | CNES code of the health facility |
| ESTABDESCR | Health facility description |

### Death Type & Date

| Field | Description |
|-------|-------------|
| TIPOBITO | Death type: 1=Fetal; 2=Non-fetal |
| DTOBITO | Date of death (ddmmyyyy) |
| HORAOBITO | Time of death |

### Deceased Information

| Field | Description |
|-------|-------------|
| NATURAL | Place of birth (1st digit=8 + state code for Brazilians) |
| CODMUNNATU | IBGE code of birth municipality |
| DTNASC | Date of birth (ddmmyyyy) |
| IDADE | Encoded age — 1st digit: 0=minutes, 1=hours, 2=days, 3=months, 4=years, 5=100+ years |
| SEXO | Sex: 0=Unknown; 1=Male; 2=Female |
| RACACOR | Race/Color: 1=White; 2=Black; 3=Asian; 4=Mixed/Brown; 5=Indigenous |
| ESTCIV | Marital status: 1=Single; 2=Married; 3=Widowed; 4=Legally separated; 5=Common-law union; 9=Unknown |
| ESC | Education: 1=None; 2=1-3 yrs; 3=4-7; 4=8-11; 5=12+; 9=Unknown |
| ESC2010 | Education (2010): 0=None; 1=Primary I; 2=Primary II; 3=Secondary; 4=Higher incomplete; 5=Higher complete; 9=Unknown |
| OCUP | Occupation (CBO-2002 — Brazilian Occupation Classification) |

### Residence & Occurrence Location

| Field | Description |
|-------|-------------|
| CODMUNRES | IBGE code of residence municipality (6 digits) |
| LOCOCOR | Place of occurrence: 1=Hospital; 2=Other health facility; 3=Home; 4=Public road; 5=Other; 6=Indigenous village; 9=Unknown |
| CODMUNOCOR | IBGE code of occurrence municipality |

### Maternal / Fetal / Perinatal Data

| Field | Description |
|-------|-------------|
| IDADEMAE | Mother's age (years) |
| ESCMAE / ESCMAE2010 | Mother's education |
| OCUPMAE | Mother's occupation (CBO-2002) |
| QTDFILVIVO | Number of living children |
| QTDFILMORT | Number of deceased children |
| GRAVIDEZ | Pregnancy type: 1=Single; 2=Twin; 3=Triplet+; 9=Unknown |
| GESTACAO / SEMAGESTAC | Gestational weeks (ranges / numeric value) |
| PARTO | Delivery type: 1=Vaginal; 2=C-section; 9=Unknown |
| OBITOPARTO | Death relative to delivery: 1=Before; 2=During; 3=After; 9=Unknown |
| PESO | Birth weight (grams) |
| OBITOGRAV | Death during pregnancy: 1=Yes; 2=No; 9=Unknown |
| OBITOPUERP | Death in puerperium: 1=0-42 days; 2=43 days-1 year; 3=No; 9=Unknown |
| TPMORTEOCO | Time of death: 1=Pregnancy; 2=Delivery; 3=Abortion; 4=Up to 42d postpartum; 5=43d-1yr; 8=Outside these; 9=Unknown |

### Medical Assistance & Confirmation

| Field | Description |
|-------|-------------|
| ASSISTMED | Medical assistance: 1=Yes; 2=No; 9=Unknown |
| EXAME | Complementary exam: 1=Yes; 2=No; 9=Unknown |
| CIRURGIA | Surgery performed: 1=Yes; 2=No; 9=Unknown |
| NECROPSIA | Autopsy performed: 1=Yes; 2=No; 9=Unknown |
| ATESTANTE | Certifier type: 1=Attending physician; 2=Substitute; 3=Forensics (IML); 4=Death Verification Service (SVO); 5=Other |

### Cause of Death

| Field | Description |
|-------|-------------|
| LINHAA / LINHAB / LINHAC / LINHAD | Certificate lines A–D (ICD-10 codes) |
| LINHAII | Part II — other significant conditions |
| CAUSABAS | **Underlying cause of death (ICD-10)** — main field |
| CAUSABAS_O | Original underlying cause (before re-selection) |
| CB_PRE | Cause selected without re-selection |
| CAUSAMAT | External cause associated with maternal death |

### External Causes

| Field | Description |
|-------|-------------|
| CIRCOBITO | Circumstance: 1=Accident; 2=Suicide; 3=Homicide; 4=Other; 9=Unknown |
| ACIDTRAB | Work accident: 1=Yes; 2=No; 9=Unknown |
| FONTE | Source: 1=Police report; 2=Hospital; 3=Family; 4=Other; 9=Unknown |

### Investigation

| Field | Description |
|-------|-------------|
| TPPOS | Investigation: 1=Investigated; 2=Not investigated |
| DTINVESTIG | Investigation date |
| FONTEINV | Investigation source (1-9) |

### Processing

| Field | Description |
|-------|-------------|
| STCODIFICA | Coded at workstation: S(Yes)/N(No) |
| CODIFICADO | Record coded: S(Yes)/N(No) |
| DTATESTADO / DTCADASTRO / DTRECEBIM / DTRECORIGA | Processing dates |
| DIFDATA | Difference (days) between death and receipt |
| STDOEPIDEM | Epidemiological death certificate: 1=Yes; 0=No |
| STDONOVA | Replacement death certificate: 1=Yes; 0=No |

```python
import pandas as pd
import numpy as np
from pysus.online_data.SIM import download
```

## 1. Fetch SIM data (sample: São Paulo, 2022)

```python
# Download mortality data (SIM-DO) for São Paulo, 2022
# groups="CID10" covers deaths coded with ICD-10
raw = download(states=["SP"], years=[2022], groups=["CID10"])
raw
```

```python
# Convert to pandas DataFrame
df = raw.to_dataframe()
print(f"Shape: {df.shape}")
df.head()
```

## 2. Basic data checks

```python
# Column types and info
df.info()
```

```python
# Null percentage per column
null_pct = (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)
print("Null % per column:\n")
print(null_pct.to_string())
```

```python
# Descriptive stats
df.describe(include="all")
```

```python
# Check key column: CAUSABAS (underlying cause of death - CID-10 code)
print(f"Unique CID-10 codes: {df['CAUSABAS'].nunique()}")
print(f"\nTop 10 causes of death:")
print(df["CAUSABAS"].value_counts().head(10))
```
